```python
import pandas as pd
import seaborn as sns

titanic = sns.load_dataset("titanic")
df = titanic.loc[0:9, ["age", "fare"]]
df
```

```python
cond = df["age"] < 20
df[cond]
```

```python
cond = df["age"] < 20
df[~cond]
```

```python
cond = (titanic["age"] >= 10) & (titanic["age"] < 20)
titanic[cond]
```

```python
# 나이가 10세 미만이고, 여성인 승객

titanic[(titanic["age"] < 10) & (titanic["sex"] == "female")]
```

```python
# 나이가 10세 미만이거나 60세 이상인 승객

cond = (titanic["age"]<10) | (titanic["age"] >=60)
titanic[cond]
```

```python
df_under10 = titanic.loc[cond,["age", "sex", "alone"]]
df_under10
```

```python
# query() 메서드 활용

# DataFrame객체.query(조건문자열)

df = titanic.loc[:9, ["age", "fare"]]
df
```

```python
df.query("age < 20")
```

```python
df.query("age >= 20 and fare < 15")
```

```python
# 나이가 20세 이상이거나 요금이 15 미만

df.query("age >= 20 or fare < 15")
```
