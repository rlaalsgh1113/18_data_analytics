```python
import pandas as pd
```

```python
dict_data = {
    'c1' : ['a', 'a', 'b', 'a', 'b'],
    'c2' : [1, 1, 1, 2, 2],
    'c3' : [1, 1, 2, 2, 2]
}
df = pd.DataFrame(dict_data)
df
```

```python
df.duplicated()
```

```python
df.duplicated(keep='last')
```

```python
df.duplicated(keep=False)
```

```python
df['c2'].duplicated()
```

```python
df.duplicated(subset=['c2'])
```

```python
# 중복 데이터 제거

df
```

```python
df2 = df.drop_duplicates()
df2
```

```python
df3 = df.drop_duplicates(keep='last')
df3
```

```python
df4 = df.drop_duplicates(keep=False)
df4
```

```python
df5 = df.drop_duplicates(subset=['c2', 'c3']) 
df5
```

```python
df.drop_duplicates(subset=['c2', 'c3'], keep=False)
```
