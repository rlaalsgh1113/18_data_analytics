```python
import pandas as pd

df = pd.read_csv("./data/auto-mpg.csv", header=None)

df.columns = [
    "mpg",
    "cylinders",
    "displacement",
    "horsepower",
    "weight",
    "acceleration",
    "model_year",
    "origin",
    "name"
]
```

```python
# 평균값

df.mean(numeric_only=True) # 숫자형 열에 대해서만 평균값을 계산함
```

```python
type(df.mean(numeric_only=True))
```

```python
# 중앙값

df.median(numeric_only=True)
```

```python
# 최댓값, 최소값
print(df.max(numeric_only=True))
print(df.min(numeric_only=True))
```

```python
print(df["origin"].max())
```

```python
# 표준편차

print(df.std(numeric_only=True))
```

```python
# 상관계수

df.corr(numeric_only=True)
```
