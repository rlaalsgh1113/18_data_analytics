```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('./data/시도별_전출입_인구수.xlsx')
df.head()
```

```python
df.info()
```

```python
df = df.ffill() # 누락값을 앞 데이터로 채움
df.head(10)
```

```python
df.tail(10)
```

```python
df.info()
```

```python
mask = (df['전출지별'] == '서울특별시') & (df['전입지별'] != '서울특별시')
df_seoul = df[mask]

df_seoul
```

```python
df_seoul = df_seoul.drop(["전출지별"], axis=1) # axis=1은 열을 의미, axis=0은 행을 의미
df_seoul
```

```python
df_seoul = df_seoul.rename({'전입지별':'전입지'}, axis=1)
df_seoul = df_seoul.set_index('전입지')
df_seoul
```

```python
sr_one = df_seoul.loc['경기도']

sr_one
```

```python
plt.plot(sr_one.index, sr_one.values)
```

```python
plt.plot(sr_one)
plt.show()
```

```python
# 차트 제목, 축 이름, 선 스타일

from matplotlib import font_manager, rc
font_path = "./data/malgun.ttf"
font_name = font_manager.FontProperties(fname=font_path).get_name()
rc('font', family= font_name)

plt.plot(sr_one.index, sr_one.values, linestyle='--')
plt.title('서울 -> 경기 인구 이동')
plt.xlabel('기간')
plt.ylabel('이동 인구수')

plt.show()
```

```python
sr_one = df_seoul.loc['경기도']

plt.figure(figsize=(14, 5))
plt.xticks(rotation="vertical")
plt.plot(sr_one.index, sr_one.values)

plt.title('서울 -> 경기 인구 이동')
plt.xlabel('기간')
plt.ylabel('이동 인구수')

plt.legend(labels=['서울 -> 경기'], loc='best') # 범례 표시
```

```python
sr_one
plt.style.use('ggplot')

plt.figure(figsize=(14, 5))
plt.xticks(rotation="vertical")
plt.plot(sr_one.index, sr_one.values,
         marker="o",
         markerfacecolor = "red",
         markeredgecolor = "blue",
         markeredgewidth = 2,
         markersize = 10)

plt.title('서울 -> 경기 인구 이동', size = 30)
plt.xlabel('기간', size = 20)
plt.ylabel('이동 인구수', size = 20)

plt.legend(labels=['서울 -> 경기'], loc='best', fontsize = 15)
```

```python
# Figure 객체, Axes 객체

import pandas as pd
import matplotlib.pyplot as plt

samsung_revenue = pd.read_csv("./data/삼성전자_분기별_매출액.csv")
samsung_revenue.head() 
samsung_revenue.info() # 데이터타입 확인
samsung_revenue = samsung_revenue.sort_values("quarter")
samsung_revenue
```

```python
fig = plt.figure()
fig
```

```python
fig, axe = plt.subplots()

print(fig)
print(axe)
```

```python
fig, axe = plt.subplots(figsize=(8,2))

axe.plot(samsung_revenue["quarter"],samsung_revenue["value"])
plt.show()
```

```python
fig, axe = plt.subplots(1,2) # 1행 2열

print(fig)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12,2))

axes[0].plot(samsung_revenue["quarter"],samsung_revenue["value"])
axes[1].plot(samsung_revenue["quarter"],samsung_revenue["value"])
samsung_revenue["value"].plot
plt.show()
```

```python
fig, axes = plt.subplots(2, 2)
print(fig)
print(axes)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 2), constrained_layout=True)
axes[0,1].plot(samsung_revenue["quarter"], samsung_revenue["value"])

plt.show()
```

```python
fig, axes = plt.subplot_mosaic(
    [["top_left", "right"],
    ["bottom_left", "right"]]
)

print(fig)
print(axes)
```

```python
fig, axes = plt.subplot_mosaic(
    [["top_left", "right"],
    ["bottom_left", "right"]],figsize=(12, 4)
)
axes["right"].plot(samsung_revenue["quarter"],samsung_revenue["value"])
plt.show()
```
